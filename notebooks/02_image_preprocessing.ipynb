```python
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image
from pathlib import Path

print("Preprocessing notebook is ready!")
```

```python
PROJECT_ROOT = Path.cwd().parent

DATASET_PATH = PROJECT_ROOT / "dataset" / "raw" / "CIFAKE"

image_path = next(
    (DATASET_PATH / "train" / "REAL").glob("*.jpg")
)

image = Image.open(image_path).convert("RGB")

image_array = np.array(image)

print("Image dimensions:", image_array.shape)
print("Data type:", image_array.dtype)
print("Minimum pixel value:", image_array.min())
print("Maximum pixel value:", image_array.max())
```

```python
normalized_image = image_array.astype(np.float32) / 255.0

print("Original range:", image_array.min(), "to", image_array.max())

print(
    "Normalized range:",
    normalized_image.min(),
    "to",
    normalized_image.max()
)

print("Normalized shape:", normalized_image.shape)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(8, 4))

axes[0].imshow(image_array)
axes[0].set_title("Original Image")
axes[0].axis("off")

axes[1].imshow(normalized_image)
axes[1].set_title("Normalized Image")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```
